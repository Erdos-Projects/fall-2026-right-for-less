```python
processed_dir = DATA_DIR / "processed"
tfidf_dir = processed_dir / "char_tfidf_50k"

questions = pd.read_pickle(
    tfidf_dir / "train_questions.pkl"
)

print("Shape:", questions.shape)
print("Columns:", questions.columns.tolist())

display(questions.head())
```

```python
scores = pd.read_pickle(
    tfidf_dir / "train_scores.pkl"
)

costs = pd.read_pickle(
    tfidf_dir / "train_costs.pkl"
)

print("Scores shape:", scores.shape)
print("Costs shape:", costs.shape)

display(scores.head())
display(costs.head())
```

```python
print(scores.columns.tolist())
```

```python
summary = pd.DataFrame({
    "Mean score": scores.mean(),
    "Mean cost": costs.mean(),
    "Cost std": costs.std(),
})

display(summary)
```

```python
from scipy.sparse import load_npz

X_tfidf = load_npz(tfidf_dir / "X_train.npz")

print("Type:", type(X_tfidf))
print("Shape:", X_tfidf.shape)
print("Nonzero entries:", X_tfidf.nnz)
```

```python
import joblib

preprocessing = joblib.load(
    tfidf_dir / "preprocessing.joblib"
)

vectorizer = preprocessing["tfidf_vectorizer"]

print("Number of TF-IDF features:",
      len(vectorizer.get_feature_names_out()))

print("First 20:")
print(vectorizer.get_feature_names_out()[:20])
```

```python
e5_dir = processed_dir / "multilingual_e5_small_chunk480"

X_e5 = np.load(e5_dir / "X_train_e5.npy")

X_e5_basic = np.load(
    e5_dir / "X_train_e5_features.npy"
)

X_e5_full = np.load(
    e5_dir / "X_train_e5_full_features.npy"
)

print("E5 only:", X_e5.shape)
print("E5 + length:", X_e5_basic.shape)
print("E5 full:", X_e5_full.shape)
```

```python
import pandas as pd
import numpy as np
from scipy.sparse import load_npz

for split in ["train", "calibration", "validation", "test"]:

    q = pd.read_pickle(tfidf_dir / f"{split}_questions.pkl")
    s = pd.read_pickle(tfidf_dir / f"{split}_scores.pkl")
    c = pd.read_pickle(tfidf_dir / f"{split}_costs.pkl")

    x_tfidf = load_npz(tfidf_dir / f"X_{split}.npz")
    x_e5 = np.load(e5_dir / f"X_{split}_e5_full_features.npy")

    print(f"\n===== {split.upper()} =====")
    print("Questions:", q.shape)
    print("Scores:   ", s.shape)
    print("Costs:    ", c.shape)
    print("TF-IDF:   ", x_tfidf.shape)
    print("E5:       ", x_e5.shape)
```
